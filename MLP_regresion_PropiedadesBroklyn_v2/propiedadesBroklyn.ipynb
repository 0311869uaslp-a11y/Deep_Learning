30. Obtenga la matriz de correlación del dataframe X y muéstrela en su formato tidy.
</div>

```python
# Matriz de correlación
# ==============================================================================
X=df.drop(['sale_price','neighborhood','tax_class','ProxCode','building_class_at_sale'],axis=1)
corr_matrix = X.corr(method='pearson')
corr_matrix
```

```python
def tidy_corr_matrix(corr_mat):
    '''
    Función para convertir una matriz de correlación de pandas en formato tidy.
    '''
    corr_mat = corr_mat.stack().reset_index()
    corr_mat.columns = ['variable_1','variable_2','r']
    corr_mat = corr_mat.loc[corr_mat['variable_1'] != corr_mat['variable_2'], :]
    corr_mat['abs_r'] = np.abs(corr_mat['r'])
    corr_mat = corr_mat.sort_values('abs_r', ascending=False)
    
    return(corr_mat)

tidy_corr_matrix(corr_matrix).head(10)
```

<div style="padding:5px;font-size:1.2em;background-color:#fdf3d7">
<h3>Evaluación del modelo</h3>
Existen varias métricas para evaluar el desempeño de los modelos de machine learning, dos de las más comunes son:
<ul>
    <li>R cuadrado $R^2$: es un coeficiente usado en el contexto de un modelo estadístico cuyo principal propósito es predecir futuros resultados o probar una hipótesis. El coeficiente determina la calidad del modelo para replicar los resultados, y la proporción de variación de los resultados que puede explicarse por el modelo.</li>
    <li>raíz del error cuadrático medio $RMSE$: es la desviación estándar de los valores residuales (errores de predicción).</li>
</ul>
</div>

```python
import numpy as np
from viznet import connecta2a, node_sequence, NodeBrush, EdgeBrush, DynamicShow

def draw_feed_forward(ax, num_node_list):
    '''
    draw a feed forward neural network.
    Args:
        num_node_list (list<int>): number of nodes in each layer.
    '''
    num_hidden_layer = len(num_node_list) - 2
    token_list = ['\sigma^z'] + \
        ['y^{(%s)}' % (i + 1) for i in range(num_hidden_layer)] + ['\psi']
    kind_list = ['nn.input'] + ['nn.hidden'] * num_hidden_layer + ['nn.output']
    radius_list = [0.3] + [0.2] * num_hidden_layer + [0.3]
    y_list = 12* np.arange(len(num_node_list))

    seq_list = []
    for n, kind, radius, y in zip(num_node_list, kind_list, radius_list, y_list):
        b = NodeBrush(kind, ax)
        seq_list.append(node_sequence(b, n, center=(0, y)))

    eb = EdgeBrush('-->', ax)
    for st, et in zip(seq_list[:-1], seq_list[1:]):
        connecta2a(st, et, eb)
def real_bp():
    with DynamicShow((30, 10), '_feed_forward.png') as d:
        draw_feed_forward(d.ax, num_node_list=[72,100,120,80, 1])

if __name__ == '__main__':
    real_bp()
```

    <li>La red neuronal dio mejor resultado (<b>33.97%</b> $R^2$) que la <b>regresión lineal múltiple</b> vista en el curso de <b>Mineria de Datos</b> que logró <b>25%</b>.</li>

Esto es normal ya que el Random Forest es un conjunto de árboles de decisión y las técnicas que combinan modelos suelen tener un mejor desempeño que el que tendrian estos de forma individual.</li>
    <li>Recuerde que para elegir un modelo no solo debe considerar su precisión, también debe tener en cuenta las características de este, por ejemplo los árboles de decisión y sus variantes (RandomForest / GradientBoosting) suelen dar un mal desempeño con valores que escapan del rango de valores de los datos de entrenamiento.</li>
    <li>En cuanto a la "transparencia del modelo" la regresión lineal y los árboles son muy claros en cuanto a que características son las más importantes al momento de realizar los calculos y tomar una decisión, lo cual no sucede con las redes neuronales. En algunas ámbitos esto suele ser importante, por ejemplo un cliente que se queje de que el banco nunca le autoriza un crédito, si se usaron redes neuronales el motivo del rechazo no quedaría muy claro.</li>
    <li>En cuanto a tiempo de entrenamiento, la regresión es muy rápida, los árboles tardan mucho, pero las redes neuronales demasiado como ya se habrá dado cuenta.</li>
    <li>La regresión lineal multiple tiene la desventaja de que la distribución de la variable objetivo debe ser parecida a la normal y que los predictores no deben tener una correlación alta, por lo que se deben aplicar técnicas de preprocesamiento para <b>outliers, correlación, VIF</b>, esto no suele ser necesario en árboles ni en redes neuronales.</li>
    <li>En cuanto a variables categóricas, al pasarlas a <b>variables dummy</b> (One hot) los árboles suelen tener problemas con demasiadas variables dummy, lo cual no es un problema para los otros 2 modelos.</li>
    <li>Por último cabe recordar que las redes neuronales tienen la capacidad de APRENDER (no solo memorizar como otros modelos) cualquier patrón que se les presente con la cantidad suficiente de neuronas y capas ocultas.</li>
