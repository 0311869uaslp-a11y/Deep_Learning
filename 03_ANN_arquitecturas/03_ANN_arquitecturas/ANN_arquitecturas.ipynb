Fuente: <a href="https://www.asimovinstitute.org/neural-network-zoo/">neural-network-zoo</a>
<img src="img/networkZooPoster.png" width="100%"/>
Aunque las arquitecturas parezcan similares entre ellas el proceso de entrenamiento y sus aplicaciones difieren bastante.
<p>Cabe señalar que, si bien la mayoría de las abreviaturas utilizadas son generalmente aceptadas, no todas lo son. Los RNN a veces se refieren a redes neuronales recursivas, pero la mayoría de las veces se refieren a redes neuronales recurrentes. Sin embargo, ese no es el final: en muchos lugares encontrará que RNN se utiliza para identificar cualquier arquitectura recurrente, incluidos LSTM, GRU e incluso las variantes bidireccionales. Los AE sufren un problema similar de vez en cuando, donde los VAE y DAE y similares se denominan simplemente AE. Muchas abreviaturas también varían en la cantidad de "N" que se agregan al final, porque se podría llamar red neuronal convolucional pero también simplemente red convolucional (lo que da como resultado CNN o CN).</p>
<p>Componer una lista completa es prácticamente imposible, ya que constantemente se inventan nuevas arquitecturas. Incluso si están publicadas, puede ser bastante difícil encontrarlas incluso si las estás buscando, a veces, simplemente pasas por alto algunas. Entonces, si bien esta lista puede brindarle algunas ideas sobre el mundo de la IA, de ninguna manera la tome por completa; especialmente si lees esta publicación mucho después de haber sido escrita (sep 2016).</p>
<ol>
    <li style="margin-top:60px;"><b>Redes neuronales prealimentadas (<i>feed forward</i>) (FF o FFNN) y los perceptrones (P o MLP) </b><br/>
    <img src="img/ff.png" width="320px" style="display:block;margin:0 auto;"/><br/>
    Son muy sencillos, alimentan información en 1 sola dirección (entrada a salida). A menudo se describe que las redes neuronales tienen capas (layer), donde cada capa consta de neuronas de entrada, ocultas o de salida en paralelo. Una capa por sí sola nunca tiene conexiones y, en general, dos capas adyacentes están completamente conectadas (cada neurona forma una conexión con cada neurona de otra capa). La red más simple y práctica tiene dos celdas de entrada y una celda de salida, que pueden usarse para modelar compuertas lógicas. Por lo general, se entrena a las FFNN mediante retropropagación (backpropagation), proporcionando a la red conjuntos de datos emparejados de "lo que entra" y "lo que queremos que salga" (aprendizaje supervisado). El error que se propaga hacia atrás suele ser alguna variación de la diferencia entre la entrada y la salida (como MSE o simplemente la diferencia lineal). Dado que la red tiene suficientes neuronas ocultas, en teoría siempre puede modelar la relación entre la entrada y la salida. Prácticamente su uso es algo limitado pero se suelen combinar con otras redes para formar nuevas redes.<br/>
    <span style="color:gray;font-size:0.8em;">Rosenblatt, Frank. “The perceptron: a probabilistic model for information storage and organization in the brain.” Psychological review 65.6 (1958): 386.</span><br/>
<a href="https://www.ling.upenn.edu/courses/cogs501/Rosenblatt1958.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Redes de función base radial (RBF)</b><br/>
    <img src="img/rbf.png" width="240px" style="display:block;margin:0 auto;"/><br/>
    Son FFNN con funciones de base radial como funciones de activación. No hay nada más. No significa que no tengan sus usos, pero la mayoría de los FFNN con otras funciones de activación no obtienen su propio nombre. Esto tiene que ver principalmente con inventarlos en el momento adecuado.<br/>
    <span style="color:gray;font-size:0.8em;">Broomhead, David S., and David Lowe. Radial basis functions, multi-variable functional interpolation and adaptive networks. No. RSRE-MEMO-4148. ROYAL SIGNALS AND RADAR ESTABLISHMENT MALVERN (UNITED KINGDOM), 1988.</span></li>
    <li style="margin-top:60px;"><b>Redes neuronales recurrentes (RNN)</b><br/>
    <img src="img/rnn.png" width="280px" style="display:block;margin:0 auto;"/><br/>
    Son FFNN con un giro temporal: tienen conexiones entre pases, conexiones a través del tiempo. Las neuronas reciben información no solo de la capa anterior sino también de ellas mismas en el paso anterior. Esto significa que <b>el orden en el que se alimenta la entrada y se entrena a la red es importante</b>: alimentarla con “leche” y luego “galletas” puede producir resultados diferentes en comparación con alimentarla con “galletas” y luego “leche”. Un gran problema con los RNN es el problema del <b>gradiente que desaparece</b> (o explota) donde, dependiendo de las funciones de activación utilizadas, la información se pierde rápidamente con el tiempo, al igual que los FFNN muy profundos pierden información en profundidad. Intuitivamente, esto no sería un gran problema porque son sólo pesos y no estados neuronales, pero los pesos a través del tiempo es en realidad donde se almacena la información del pasado; si el peso alcanza un valor de 0 o 1.000.000, el estado anterior no será muy informativo. En principio, los RNN se pueden usar en muchos campos, ya que la mayoría de las formas de datos que en realidad no tienen una línea de tiempo (es decir, a diferencia del sonido o el video) se pueden representar como una secuencia. Una imagen o una cadena de texto se puede alimentar un píxel o carácter a la vez, por lo que los pesos dependientes del tiempo se utilizan para lo que vino antes en la secuencia, no para lo que sucedió x segundos antes. En general, las redes recurrentes son una buena opción para avanzar o completar información, como el autocompletado.<br/>
    <span style="color:gray">Elman, Jeffrey L. “Finding structure in time.” Cognitive science 14.2 (1990): 179-211.</span></li>
    <li style="margin-top:60px;"><b>Las redes de memoria a largo y corto plazo (LSTM)</b><br/>
    <img src="img/lstm.png" width="280px" style="display:block;margin:0 auto;"/><br/>
    Intentan combatir el problema del gradiente que desaparece o explota introduciendo puertas y una celda de memoria definida explícitamente. Estos están inspirados principalmente en circuitos, no tanto en biología. <b>Cada neurona tiene una celda de memoria y tres puertas</b>: entrada, salida y olvido. La función de estas puertas es salvaguardar la información deteniendo o permitiendo el flujo de la misma. La puerta de entrada determina cuánta información de la capa anterior se almacena en la celda. La capa de salida toma el trabajo en el otro extremo y determina qué parte de la siguiente capa llega a conocer el estado de esta celda. La puerta de olvido parece una inclusión extraña al principio, pero a veces es bueno olvidar: si estás aprendiendo un libro y comienza un nuevo capítulo, puede ser necesario que la red olvide algunos personajes del capítulo anterior. Se ha demostrado que los LSTM pueden aprender secuencias complejas, como escribir como Shakespeare o componer música primitiva. Tenga en cuenta que cada una de estas puertas tiene un peso para una célula de la neurona anterior, por lo que normalmente requieren más recursos para funcionar.<br/>
    <span style="color:gray;font-size:0.8em;">Hochreiter, Sepp, and Jürgen Schmidhuber. “Long short-term memory.” Neural computation 9.8 (1997): 1735-1780.</span><br/>
    <a href="https://www.bioinf.jku.at/publications/older/2604.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las unidades recurrentes cerradas (GRU) (Gated recurrent units)</b><br/>
    <img src="img/gru.png" width="280px" style="display:block;margin:0 auto;"/><br/>
    Son una ligera variación de los LSTM. Tienen una puerta menos y están conectados de forma ligeramente diferente: en lugar de una puerta de entrada, salida y olvido, tienen una <b>puerta de actualización</b>. Esta puerta de actualización determina cuánta información conservar del último estado y cuánta información dejar entrar de la capa anterior. La <b>puerta de reinicio</b> funciona de manera muy similar a la puerta de olvido de un LSTM, pero está ubicada de manera ligeramente diferente. Siempre envían su estado completo, no tienen puerta de salida. En la mayoría de los casos, funcionan de manera muy similar a los LSTM, con la mayor diferencia que los GRU son un poco más rápidos y más fáciles de ejecutar (pero también un poco menos expresivos). En la práctica, estos tienden a anularse entre sí, ya que se necesita una red más grande para recuperar algo de expresividad, lo que a su vez anula los beneficios de rendimiento. En algunos casos en los que no se necesita expresividad adicional, los GRU pueden superar a los LSTM.<br/>
    <span style="color:gray;font-size:0.8em;">Chung, Junyoung, et al. “Empirical evaluation of gated recurrent neural networks on sequence modeling.” arXiv preprint arXiv:1412.3555 (2014).</span><br/>
    <a href="https://arxiv.org/pdf/1412.3555v1.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las redes neuronales recurrentes bidireccionales, las redes de memoria bidireccionales a largo y corto plazo y las unidades recurrentes cerradas bidireccionales (BiRNN, BiLSTM y BiGRU respectivamente) </b><br/>
    No se muestran en el gráfico porque se ven exactamente iguales que sus contrapartes unidireccionales. La diferencia es que estas redes no sólo están conectadas con el pasado, sino también con el futuro. Como ejemplo, los LSTM unidireccionales podrían entrenarse para predecir la palabra "pez" al recibir las letras una por una, donde las conexiones recurrentes a lo largo del tiempo recuerdan el último valor. Un BiLSTM también recibiría la siguiente letra de la secuencia en el pase hacia atrás, dándole acceso a información futura. Esto entrena a la red para llenar vacíos en lugar de hacer avanzar información, por lo que en lugar de expandir una imagen en el borde, podría llenar un agujero en el medio de una imagen.<br/>
     <span style="color:gray;font-size:0.8em;">Schuster, Mike y Kuldip K. Paliwal. "Redes neuronales recurrentes bidireccionales". Transacciones IEEE sobre procesamiento de señales 45.11 (1997): 2673-2681.</span></li>
    <li style="margin-top:60px;"><b>Los codificadores automáticos (AE)</b><br/>
    <img src="img/ae.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Son algo similares a los FFNN, ya que los AE se parecen más a un uso diferente de los FFNN que a una arquitectura fundamentalmente diferente. La idea básica detrás de los codificadores automáticos es codificar información (como comprimir, no cifrar) automáticamente, de ahí el nombre. Toda la red siempre tiene <b>forma de reloj de arena</b>, con capas ocultas más pequeñas que las capas de entrada y salida. Los AE también son siempre simétricos alrededor de las capas intermedias (una o dos dependiendo de la cantidad par o impar de capas). Las capas más pequeñas están casi siempre en el medio, el lugar donde la información está más comprimida (el punto de cuello de la red). Todo lo que está hasta el medio se llama parte de codificación, todo lo que está después del medio es la decodificación y el medio (sorpresa) el código. Se pueden entrenar usando la retropropagación alimentando la entrada y configurando el error como la diferencia entre la entrada y lo que salió. Los AE también se pueden construir simétricamente en lo que respecta a los pesos, por lo que los pesos de codificación son los mismos que los de decodificación.<br/>
    <span style="color:gray;font-size:0.8em;">Bourlard, Hervé, and Yves Kamp. “Auto-association by multilayer perceptrons and singular value decomposition.” Biological cybernetics 59.4-5 (1988): 291-294.</span><br/>
    <a href="https://www.semanticscholar.org/paper/Auto-association-by-multilayer-perceptrons-and-Bourlard-Kamp/f5821548720901c89b3b7481f7500d7cd64e99bd?p2df">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Los codificadores automáticos variacionales (VAE)</b><br/>
    <img src="img/vae.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Tienen la misma arquitectura que los AE, pero se les "enseña" algo más: una distribución de probabilidad aproximada de las muestras de entrada. Es un poco volver a las raíces, ya que están un poco más estrechamente relacionados con los BMs y los RBMs. Sin embargo, se basan en las matemáticas bayesianas con respecto a la inferencia probabilística y la independencia, así como en un truco de reparametrización para lograr esta representación diferente. Las partes de inferencia e independencia tienen sentido intuitivamente, pero se basan en matemáticas algo complejas. Lo básico se reduce a esto: tener en cuenta la influencia. Si una cosa sucede en un lugar y otra sucede en otro lugar, no necesariamente están relacionadas. Si no están relacionados, entonces la propagación del error debería tenerlo en cuenta. Este es un enfoque útil porque las redes neuronales son gráficos grandes (en cierto modo), por lo que ayuda si puedes descartar la influencia de algunos nodos a otros nodos a medida que te sumerges en capas más profundas.<br/>
    <span style="color:gray;font-size:0.8em;">Kingma, Diederik P., and Max Welling. “Auto-encoding variational bayes.” arXiv preprint arXiv:1312.6114 (2013).</span><br/>
    <a href="https://arxiv.org/pdf/1312.6114v10.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Los codificadores automáticos de eliminación de ruido (DAE) (Denoising AE)</b><br/>
    <img src="img/dae.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Son AE en los que no alimentamos solo los datos de entrada, sino que los alimentamos con ruido (como hacer una imagen más granulada). Sin embargo, calculamos el error de la misma manera, de modo que la salida de la red se compara con la entrada original sin ruido. Esto anima a la red a <B>no aprender detalles sino características más amplias</B>, ya que aprender características más pequeñas a menudo resulta "incorrecto" debido a que cambia constantemente con el ruido.<br/>
    <span style="color:gray;font-size:0.8em;">Vincent, Pascal, et al. “Extracting and composing robust features with denoising autoencoders.” Proceedings of the 25th international conference on Machine learning. ACM, 2008.</span><br/>
    <a href="http://machinelearning.org/archive/icml2008/papers/592.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Los codificadores automáticos dispersos (SAE) (sparse ae)</b><br/>
    <img src="img/sae.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Son, en cierto modo, lo opuesto a los AE. En lugar de enseñarle a una red a representar un montón de información en menos "espacio" o nodos, intentamos codificar información en más espacio. Entonces, en lugar de que la red converja en el medio y luego se expanda nuevamente al tamaño de entrada, explotamos el medio. Estos tipos de redes se pueden utilizar para extraer muchas características pequeñas de un conjunto de datos. Si uno entrenara un SAE de la misma manera que un AE, en casi todos los casos terminaría con una red de identidad bastante inútil (ya que lo que entra es lo que sale, sin ninguna transformación o descomposición). Para evitar esto, en lugar de retroalimentar la entrada, retroalimentamos la entrada más un controlador de dispersión. Este controlador de escasez puede tomar la forma de un filtro de umbral, donde solo se devuelve y entrena un determinado error, el otro error será "irrelevante" para ese pase y se establecerá en cero. En cierto modo, esto se parece a las redes neuronales con picos, donde no todas las neuronas se activan todo el tiempo (y se puntúan puntos por la plausibilidad biológica).<br/>
     <span style="color:gray;font-size:0.8em;">Marc’Aurelio Ranzato, Christopher Poultney, Sumit Chopra, and Yann LeCun. “Efficient learning of sparse representations with an energy-based model.” Proceedings of NIPS. 2007.</span><br/>
     <a href="https://proceedings.neurips.cc/paper_files/paper/2006/file/87f4d79e36d68c3031ccf6c55e9bbd39-Paper.pdf" >Artículo original PDF</a></li><br/>
    <div style="margin-top:60px;"><b>Las cadenas de Markov (MC o cadena de Markov de tiempo discreto, DTMC)</b><br/>
    <img src="img/mc.png" width="240px" style="display:block;margin:0 auto;"/><br/>
    Son una especie de predecesoras de las <b>BM</b> y <b>HN</b>. Se pueden entender de la siguiente manera: desde este nodo en el que estoy ahora, ¿cuáles son las probabilidades de que vaya a cualquiera de mis nodos vecinos? No tienen memoria (propiedad de Markov), lo que significa que cada estado en el que terminas depende completamente del estado anterior. Si bien no son realmente una red neuronal, se parecen a las redes neuronales y forman la base teórica para los BM y HN. Las MC no siempre se consideran redes neuronales, como ocurre con las BM, RBM y HN. Las cadenas de Markov tampoco siempre están completamente conectadas.<br/>
    <span style="color:gray;font-size:0.8em;">Hayes, Brian. “First links in the Markov chain.” American Scientist 101.2 (2013): 252.</span><br/>
    <a href="https://www.americanscientist.org/sites/americanscientist.org/files/201321152149545-2013-03Hayes.pdf">Artículo original PDF</a>
    </div>
    <li style="margin-top:60px;"><b>Red Hopfield (HN) </b><br/>
    <img src="img/hn.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Es una red donde cada neurona está conectada a todas las demás neuronas; es un <b>"plato de espaguetis completamente enredado"</b> ya que incluso todos los nodos funcionan como todo. Cada nodo se ingresa antes del entrenamiento, luego se oculta durante el entrenamiento y se genera después. Las redes se entrenan estableciendo el valor de las neuronas en el patrón deseado, después del cual se pueden calcular los pesos. Los pesos no cambian después de esto. Una vez entrenada para uno o más patrones, la red siempre convergerá a uno de los patrones aprendidos porque la red solo es estable en esos estados. Tenga en cuenta que no siempre se ajusta al estado deseado (lamentablemente no es una caja negra mágica). Se estabiliza en parte debido a que la "energía" o "temperatura" total de la red se reduce gradualmente durante el entrenamiento. Cada neurona tiene un umbral de activación que escala hasta esta temperatura, que si se supera sumando las entradas hace que la neurona adopte la forma de uno de dos estados (normalmente -1 o 1, a veces 0 o 1). La actualización de la red se puede realizar de forma sincrónica o, más comúnmente, una por una. Si se actualiza una por una, se crea una secuencia aleatoria justa para organizar qué celdas se actualizan en qué orden (siendo aleatorias justas todas las opciones (n) que ocurren exactamente una vez cada n elementos). Esto es para que pueda saber cuándo la red es estable (ha terminado de converger), una vez que cada celda se ha actualizado y ninguna de ellas ha cambiado, la red es estable (recocida). Estas redes a menudo se denominan memoria asociativa porque convergen al estado más similar al de la entrada; Si los humanos ven media mesa, podemos visualizar la otra mitad; esta red convergerá en una mesa si se presenta con la mitad de ruido y media mesa.<br/>
    <span style="color:gray;font-size:0.8em;">Hopfield, John J. “Neural networks and physical systems with emergent collective computational abilities.” Proceedings of the national academy of sciences 79.8 (1982): 2554-2558.</span><br/>
    <a href="https://bi.snu.ac.kr/Courses/g-ai09-2/hopfield82.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las máquinas de Boltzmann (BM) </b><br/>
    <img src="img/bm.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Se parecen mucho a las HN, pero algunas neuronas están marcadas como neuronas de entrada y otras permanecen "ocultas". Las neuronas de entrada se convierten en neuronas de salida al final de una actualización completa de la red. Comienza con pesos aleatorios y aprende mediante <span style="color:red">retropropagación</span> o, más recientemente, mediante <span style="color:red">divergencia contrastiva</span> (se utiliza una cadena de Markov para determinar los gradientes entre dos ganancias de información). En comparación con una HN, las neuronas tienen en su mayoría patrones de activación binarios. Como lo sugiere el hecho de haber sido entrenados por MC, los BM son <b>redes estocásticas</b>. El proceso de entrenamiento y ejecución de un BM es bastante similar al de un HN: se configuran las neuronas de entrada a ciertos valores fijados, después de lo cual la red se libera. Mientras están libres, las células pueden obtener cualquier valor y repetidamente vamos y venimos entre las neuronas de entrada y las ocultas. La activación está controlada por un valor de temperatura global, que si se reduce disminuye la energía de las células. Esta menor energía hace que sus patrones de activación se estabilicen. La red alcanza un equilibrio dada la temperatura adecuada.<br/>
    <span style="color:gray;font-size:0.8em;">Hinton, Geoffrey E., and Terrence J. Sejnowski. “Learning and releaming in Boltzmann machines.” Parallel distributed processing: Explorations in the microstructure of cognition 1 (1986): 282-317.</span><br/>
    <a href="https://www.researchgate.net/publication/242509302_Learning_and_relearning_in_Boltzmann_machines">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las máquinas Boltzmann restringidas (RBM)</b><br/>
    <img src="img/rbm.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Son notablemente similares a las <b>BM</b> y, por tanto, también a las <b>HN</b>. La mayor diferencia entre BM y RBM es que los <b>RBM son más utilizables porque son más restringidos</b>. No conectan cada neurona con cada otra neurona, sino que solo conectan cada grupo diferente de neuronas con otro grupo, por lo que ninguna neurona de entrada está conectada directamente a otras neuronas de entrada y tampoco se realizan conexiones de ocultas a ocultas. Los RBM se pueden entrenar como <b>FFNN</b> con un cambio: en lugar de pasar datos hacia adelante y luego propagarlos hacia atrás, se pasan los datos hacia adelante y luego se pasan los datos hacia atrás (de regreso a la primera capa). Después de eso, entrenas con propagación hacia adelante y hacia atrás.<br/>
    <span style="color:gray;font-size:0.8em;">Smolensky, Paul. Information processing in dynamical systems: Foundations of harmony theory. No. CU-CS-321-86. COLORADO UNIV AT BOULDER DEPT OF COMPUTER SCIENCE, 1986.</span><br/></li>
    <li style="margin-top:60px;"><b>Redes de creencia profunda (DBN) (Deep belief)</b><br/>
    <img src="img/dbn.png" width="420px" style="display:block;margin:0 auto;"/><br/>
    Es el nombre que se le da a las arquitecturas apiladas, en su mayoría <b>RBM</b> o <b>VAE</b>. Se ha demostrado que estas redes se pueden entrenar de manera efectiva pila por pila, donde cada AE o RBM solo tiene que aprender a codificar la red anterior. Esta técnica también se conoce como entrenamiento codicioso, donde codicioso significa crear <b>soluciones localmente óptimas para llegar a una respuesta decente pero posiblemente no óptima</b>. Los DBN se pueden entrenar mediante divergencia contrastiva o backpropagation y aprender a representar los datos como un modelo probabilístico, al igual que los RBM o VAE normales. Una vez entrenado o convergido a un estado (más) estable mediante aprendizaje no supervisado, el modelo se puede utilizar para generar nuevos datos. Si se entrena con divergencia contrastiva, puede incluso clasificar datos existentes porque a las neuronas se les ha enseñado a buscar características diferentes.<br/>
    Antes de las DBN en 2006 los modelos con “profundidad” (decenas o cientos de capas) eran considerados demasiado difíciles de entrenar (incluso con backpropagation) y el uso de las redes neuronales artificiales quedó estancado. Con la creación de una DBN que logro obtener un mejor resultado en el <b>MNIST</b>, se devolvió el entusiasmo en poder lograr el aprendizaje profundo en redes neuronales. Hoy en día las DBN no se utilizan demasiado, pero fueron un gran hito en la historia en el desarrollo del deep learning y permitieron seguir la exploración para mejorar las redes existentes CNN, LSTM, etc.<br/>
Las Deep Belief Networks, demostraron que utilizar pesos aleatorios al inicializar las redes son una mala idea: por ejemplo al utilizar Backpropagation con Descenso por gradiente muchas veces se caía en mínimos locales, sin lograr optimizar los pesos. Mejor será utilizar una asignación de pesos inteligente mediante un preentrenamiento de las capas de la red -en inglés “pretrain”-. Se basa en el uso de la utilización de Restricted Boltzmann Machines y Autoencoders para pre-entrenar la red de manera no supervisada. Ojo! luego de pre-entrenar y asignar esos pesos iniciales, deberemos entrenar la red por de forma habitual, supervisada (por ejemplo con backpropagation).<br/>
    <span style="color:gray;font-size:0.8em;">Bengio, Yoshua, et al. “Greedy layer-wise training of deep networks.” Advances in neural information processing systems 19 (2007): 153.</span><br/>
    <a href="https://proceedings.neurips.cc/paper_files/paper/2006/file/5da713a690c067105aeb2fae32403405-Paper.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las redes neuronales convolucionales (CNN o redes neuronales convolucionales profundas, DCNN)</b><br/>
    <img src="img/cnn.png" width="450px" style="display:block;margin:0 auto;"/><br/>
    Son bastante diferentes de la mayoría de las demás redes. Se utilizan principalmente para el <b>procesamiento de imágenes</b>, pero también se pueden utilizar para otros tipos de entrada, como audio. Un caso de uso típico para las CNN es cuando alimentas la red con imágenes y la red clasifica los datos, por ejemplo, genera "gato" si le das una imagen de gato y "perro" cuando le das una imagen de perro. Las CNN tienden a comenzar con un "escáner" de entrada que no pretende analizar todos los datos de entrenamiento a la vez. Por ejemplo, para ingresar una imagen de 200 x 200 píxeles, no querrás una capa con 40 000 nodos. Más bien, crea una capa de entrada de escaneo de, digamos, 20 x 20 a la que alimenta los primeros 20 x 20 píxeles de la imagen (generalmente comenzando en la esquina superior izquierda). Una vez que haya pasado esa entrada (y posiblemente la use para entrenamiento), la alimentará con los siguientes 20 x 20 píxeles: moverá el escáner un píxel hacia la derecha. Tenga en cuenta que uno no movería la entrada 20 píxeles (o cualquier ancho del escáner), no está diseccionando la imagen en bloques de 20 x 20, sino que se está arrastrando sobre ella. Estos datos de entrada luego se alimentan a través de capas convolucionales en lugar de capas normales, donde no todos los nodos están conectados a todos los nodos. Cada nodo solo se ocupa de las celdas vecinas cercanas (cuán cerca depende de la implementación, pero generalmente no más de unas pocas). Estas capas convolucionales también tienden a reducirse a medida que se vuelven más profundas, principalmente por factores de entrada fácilmente divisibles (por lo que 20 probablemente irían a una capa de 10 seguida de una capa de 5). Las potencias de dos se utilizan con mucha frecuencia aquí, ya que se pueden dividir limpia y completamente por definición: 32, 16, 8, 4, 2, 1. Además de estas capas convolucionales, también suelen presentar capas de agrupación. La agrupación es una forma de filtrar detalles: una técnica de agrupación común es la agrupación máxima, donde tomamos, digamos, 2 x 2 píxeles y pasamos el píxel con la mayor cantidad de rojo. Para aplicar CNN para audio, básicamente alimenta las ondas de audio de entrada y avanza lentamente a lo largo del clip, segmento por segmento. Las implementaciones de CNN en el mundo real a menudo pegan una <b>FFNN</b> al final para procesar aún más los datos, lo que permite abstracciones altamente no lineales. Estas redes se denominan DCNN, pero los nombres y abreviaturas entre estas dos suelen utilizarse indistintamente.<br/>
      <span style="color:gray;font-size:0.8em;">LeCun, Yann, et al. “Gradient-based learning applied to document recognition.” Proceedings of the IEEE 86.11 (1998): 2278-2324.</span><br/>
      <a href="http://yann.lecun.com/exdb/publis/pdf/lecun-98.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las redes deconvolucionales (DN)</b><br/>
    <img src="img/dn.png" width="450px" style="display:block;margin:0 auto;"/><br/>
    También llamadas redes de gráficos inversos (<b>IGN</b>), son redes neuronales convolucionales invertidas. Imagine alimentar a una red con la palabra "gato" y entrenarla para que produzca imágenes felinas, comparando lo que genera con imágenes reales de gatos. Las DNN se pueden combinar con las <b>FFNN</b> como las <b>CNN</b> normales, pero este es el punto en el que se traza la línea con la creación de nuevas abreviaturas. Se puede hacer referencia a ellas como redes neuronales deconvolucionales profundas, pero se podría argumentar que cuando coloca FFNN en la parte posterior y frontal de las DNN, tiene otra arquitectura que merece un nuevo nombre. Tenga en cuenta que en la mayoría de las aplicaciones uno no alimentaría la red con entradas similares a texto, más bien con un vector de entrada de clasificación binaria. Piensa que $<0, 1>$ es un gato, $<1, 0>$ es un perro y $<1, 1>$ es un gato y un perro. Las capas de agrupación que se encuentran comúnmente en las CNN a menudo se reemplazan con operaciones inversas similares, principalmente interpolación y extrapolación con suposiciones sesgadas (si una capa de agrupación usa agrupación máxima, puede inventar datos nuevos exclusivamente más bajos al revertirla).<br/>
      <span style="color:gray;font-size:0.8em;">Zeiler, Matthew D., et al. “Deconvolutional networks.” Computer Vision and Pattern Recognition (CVPR), 2010 IEEE Conference on. IEEE, 2010.</span><br/>
      <a href="https://www.matthewzeiler.com/mattzeiler/deconvolutionalnetworks.pdf">Artículo original PDF</a></li>
      <li style="margin-top:60px;"><b>Las redes de gráficos inversos convolucionales profundos (DCIGN)</b><br/>
    <img src="img/dcign.png" width="420px" style="display:block;margin:0 auto;"/><br/>
    Tienen un nombre un tanto engañoso, ya que en realidad son <b>VAE</b> pero con <b>CNN</b> y <b>DNN</b> para los respectivos codificadores y decodificadores. Estas redes intentan modelar “características” en la codificación como probabilidades, de modo que puedan aprender a producir una imagen con un gato y un perro juntos, habiendo visto solo uno de los dos en imágenes separadas. Del mismo modo, podrías darle una foto de un gato con el molesto perro de tus vecinos y pedirle que elimine al perro, sin haber realizado nunca dicha operación. Las pruebas han demostrado que estas redes también pueden aprender a modelar transformaciones complejas en imágenes, como cambiar la fuente de luz o la rotación de un objeto 3D. Estas redes tienden a entrenarse con <b>backpropagation</b>.<br/>
      <span style="color:gray;font-size:0.8em;">Kulkarni, Tejas D., et al. “Deep convolutional inverse graphics network.” Advances in Neural Information Processing Systems. 2015.</span><br/>
    <a href="https://arxiv.org/pdf/1503.03167v4.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las redes generativas adversarias (GAN)</b><br/>
    <img src="img/gan.png" width="420px" style="display:block;margin:0 auto;"/><br/>
          pertenecen a un tipo diferente de redes, son gemelas: dos redes que trabajan juntas. Las GAN constan de dos redes cualesquiera (aunque a menudo son una combinación de FF y CNN), una de las cuales tiene la tarea de generar contenido y la otra tiene que juzgar el contenido. La red discriminante recibe datos de entrenamiento o contenido generado de la red generativa. La eficacia con la que la red discriminadora pudo predecir correctamente la fuente de datos se utiliza como parte del error de la red generadora. Esto crea una forma de competencia en la que el discriminador mejora en distinguir los datos reales de los datos generados y el generador aprende a ser menos predecible para el discriminador. Esto funciona bien en parte porque incluso los patrones similares a ruidos bastante complejos son eventualmente predecibles, pero el contenido generado similar en características a los datos de entrada es más difícil de aprender a distinguir. Las GAN pueden ser bastante difíciles de entrenar, ya que no solo es necesario entrenar dos redes (cualquiera de las cuales puede plantear sus propios problemas), sino que también es necesario equilibrar su dinámica. Si la predicción o la generación son demasiado buenas en comparación con la otra, una GAN no convergerá porque existe una divergencia intrínseca.
    <span style="color:gray;font-size:0.8em;">Goodfellow, Ian, et al. “Generative adversarial nets.” Advances in Neural Information Processing Systems (2014).</span><br/>
    <a href="https://arxiv.org/pdf/1406.2661v1.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las máquinas de estado líquido (LSM)</b><br/>
    <img src="img/lsm.png" width="420px" style="display:block;margin:0 auto;"/><br/>  
        son sopas similares y se parecen mucho a las ESN. La verdadera diferencia es que los LSM son un tipo de redes neuronales de picos: las activaciones sigmoideas se reemplazan con funciones de umbral y cada neurona es también una célula de memoria que se acumula. Entonces, al actualizar una neurona, el valor no se establece en la suma de las vecinas, sino que se suma a sí misma. Una vez que se alcanza el umbral, libera su energía a otras neuronas. Esto crea un patrón similar a un pico, donde no sucede nada durante un tiempo hasta que de repente se alcanza un umbral.
    <span style="color:gray;font-size:0.8em;">Maass, Wolfgang, Thomas Natschläger, and Henry Markram. “Real-time computing without stable states: A new framework for neural computation based on perturbations.” Neural computation 14.11 (2002): 2531-2560.</span><br/>
    <a href="https://web.archive.org/web/20120222154641/http://ramsesii.upf.es/seminar/Maass_et_al_2002.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las máquinas de aprendizaje extremo (ELM)</b><br/>
    <img src="img/elm.png" width="420px" style="display:block;margin:0 auto;"/><br/>  
    son básicamente FFNN pero con conexiones aleatorias. Se parecen mucho a los LSM y ESN, pero no son recurrentes ni tienen picos. Tampoco utilizan retropropagación. En cambio, comienzan con pesos aleatorios y entrenan los pesos en un solo paso de acuerdo con el ajuste de mínimos cuadrados (el error más bajo en todas las funciones). Esto da como resultado una red mucho menos expresiva pero también es mucho más rápida que la retropropagación.
    <span style="color:gray;font-size:0.8em;">Huang, Guang-Bin, et al. “Extreme learning machine: Theory and applications.” Neurocomputing 70.1-3 (2006): 489-501.</span><br/>
    <a href="https://www.sciencedirect.com/science/article/pii/S0925231206000385?via%3Dihub">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las redes de estado de eco (ESN)</b><br/>
    <img src="img/esn.png" width="420px" style="display:block;margin:0 auto;"/><br/>    
    son otro tipo diferente de red (recurrente). Éste se distingue de los demás por tener conexiones aleatorias entre las neuronas (es decir, no organizadas en conjuntos ordenados de capas) y están entrenadas de manera diferente. En lugar de alimentar la entrada y propagar el error hacia atrás, alimentamos la entrada, la reenviamos y actualizamos las neuronas por un tiempo, y observamos la salida a lo largo del tiempo. Las capas de entrada y salida tienen un papel poco convencional, ya que la capa de entrada se utiliza para preparar la red y la capa de salida actúa como observadora de los patrones de activación que se desarrollan con el tiempo. Durante el entrenamiento, sólo se cambian las conexiones entre el observador y las (sopa de) unidades ocultas.
    <span style="color:gray;font-size:0.8em;">Jaeger, Herbert, and Harald Haas. “Harnessing nonlinearity: Predicting chaotic systems and saving energy in wireless communication.” science 304.5667 (2004): 78-80.</span><br/>
    <a href="http://science.sciencemag.org/content/304/5667/78/tab-pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las redes residuales profundas (DRN)</b><br/>
    <img src="img/drn.png" width="420px" style="display:block;margin:0 auto;"/><br/>  
    son FFNN muy profundas con conexiones adicionales que pasan información de una capa a una capa posterior (a menudo de 2 a 5 capas), así como a la siguiente capa. En lugar de intentar encontrar una solución para asignar alguna entrada a alguna salida a través de, digamos, 5 capas, se obliga a la red a aprender a asignar alguna entrada a alguna salida + alguna entrada. Básicamente, agrega una identidad a la solución, trasladando la entrada anterior y entregándola nuevamente a una capa posterior. Se ha demostrado que estas redes son muy efectivas para aprender patrones de hasta 150 capas de profundidad, mucho más que las 2 a 5 capas habituales que uno podría esperar entrenar. Sin embargo, se ha demostrado que estas redes son, en esencia, solo RNN sin la construcción explícita basada en el tiempo y, a menudo, se comparan con LSTM sin puertas.
    <span style="color:gray;font-size:0.8em;">He, Kaiming, et al. “Deep residual learning for image recognition.” arXiv preprint arXiv:1512.03385 (2015).</span><br/>
    <a href="https://arxiv.org/pdf/1512.03385v1.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las máquinas neuronales de Turing (NTM)</b><br/>
    <img src="img/ntm.png" width="420px" style="display:block;margin:0 auto;"/><br/> 
    pueden entenderse como una abstracción de las LSTM y un intento de eliminar las redes neuronales de la caja negra (y darnos una idea de lo que sucede allí). En lugar de codificar una célula de memoria directamente en una neurona, la memoria se separa. Es un intento de combinar la eficiencia y permanencia del almacenamiento digital regular y la eficiencia y el poder expresivo de las redes neuronales. La idea es tener un banco de memoria direccionable por contenido y una red neuronal que pueda leer y escribir en él. El "Turing" en las máquinas neuronales de Turing proviene de que son Turing completo: la capacidad de leer, escribir y cambiar de estado en función de lo que lee significa que puede representar cualquier cosa que una máquina universal de Turing pueda representar.
    <span style="color:gray;font-size:0.8em;">Graves, Alex, Greg Wayne, and Ivo Danihelka. “Neural turing machines.” arXiv preprint arXiv:1410.5401 (2014).</span><br/>
    <a href="https://arxiv.org/pdf/1410.5401v2.pdf">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Las computadoras neuronales diferenciables (DNC) </b><br/>
    <img src="img/dmc.png" width="380px" style="display:block;margin:0 auto;"/><br/>  
    son máquinas de Turing neuronales mejoradas con memoria escalable, inspiradas en cómo el hipocampo humano almacena los recuerdos. La idea es tomar la arquitectura informática clásica de Von Neumann y reemplazar la CPU con un RNN, que aprende cuándo y qué leer de la RAM. Además de tener un gran banco de números como memoria (que se puede cambiar de tamaño sin volver a entrenar el RNN). La DNC también cuenta con tres mecanismos de atención. Estos mecanismos permiten al RNN consultar la similitud de un bit de entrada con las entradas de la memoria, la relación temporal entre dos entradas cualesquiera en la memoria y si una entrada de la memoria se actualizó recientemente, lo que hace que sea menos probable que se sobrescriba cuando no hay ninguna entrada vacía. memoria disponible.
    <span style="color:gray;font-size:0.8em;">Graves, Alex, et al. “Hybrid computing using a neural network with dynamic external memory.” Nature 538 (2016): 471-476.</span><br/>
    <a href="https://www.nature.com/articles/nature20101">Artículo original PDF</a></li>
    <li style="margin-top:60px;"><b>Capsule Networks (CapsNet)</b><br/>
    <img src="img/cn.png" width="420px" style="display:block;margin:0 auto;"/><br/>  
    son alternativas a la agrupación inspiradas en la biología, donde las neuronas están conectadas con múltiples pesos (un vector) en lugar de solo un peso (un escalar). Esto permite a las neuronas transferir más información que simplemente qué característica se detectó, como dónde se encuentra una característica en la imagen o qué color y orientación tiene. El proceso de aprendizaje implica una forma local de aprendizaje hebbiano que valora las predicciones correctas de resultados en la siguiente capa.
    <span style="color:gray;font-size:0.8em;">Sabour, Sara, Frosst, Nicholas, and Hinton, G. E. “Dynamic Routing Between Capsules.” In Advances in neural information processing systems (2017): 3856-3866.</span><br/>
    <a href="https://arxiv.org/abs/1710.09829">Artículo original PDF</a></li>   
    <li style="margin-top:60px;"><b>Las redes Kohonen (KN, también mapa (de funciones) autoorganizado, SOM, SOFM)</b><br/>
    <img src="img/kn.png" width="300px" style="display:block;margin:0 auto;"/><br/>   
     utilizan el aprendizaje competitivo para clasificar datos sin supervisión. La entrada se presenta a la red, después de lo cual la red evalúa cuál de sus neuronas coincide más con esa entrada. Luego, estas neuronas se ajustan para que coincidan aún mejor con la entrada, arrastrando a sus vecinas en el proceso. Cuánto se mueven los vecinos depende de la distancia de los vecinos a las mejores unidades coincidentes.
    <span style="color:gray;font-size:0.8em;">Kohonen, Teuvo. “Self-organized formation of topologically correct feature maps.” Biological cybernetics 43.1 (1982): 59-69.</span><br/>
    <a href="http://cioslab.vcu.edu/alg/Visualize/kohonen-82.pdf">Artículo original PDF</a></li>  
    <li style="margin-top:60px;"><b>Las redes de atención (AN)</b><br/>
    <img src="img/an.png" width="420px" style="display:block;margin:0 auto;"/><br/>  
    pueden considerarse una clase de redes, que incluye la arquitectura Transformer. Utilizan un mecanismo de atención para combatir el deterioro de la información almacenando por separado los estados anteriores de la red y cambiando la atención entre los estados. Los estados ocultos de cada iteración en las capas de codificación se almacenan en celdas de memoria. Las capas de decodificación están conectadas a las capas de codificación, pero también recibe datos de las celdas de memoria filtrados por un contexto de atención. Este paso de filtrado agrega contexto para las capas de decodificación enfatizando la importancia de características particulares. La red de atención que produce este contexto se entrena utilizando la señal de error de la salida de la capa de decodificación. Además, el contexto de atención se puede visualizar brindando información valiosa sobre qué características de entrada se corresponden con qué características de salida.
    <span style="color:gray;font-size:0.8em;">Jaderberg, Max, et al. “Spatial Transformer Networks.” In Advances in neural information processing systems (2015): 2017-2025.</span><br/>
    <a href="https://arxiv.org/pdf/1506.02025.pdf">Artículo original PDF</a></li>
