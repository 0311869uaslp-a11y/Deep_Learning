<h1 style="color:blue;margin-top:2px;">Redes neuronales</h1>

<h2>Introducción</h2>
Una RNA (Red Neuronal Artificial) es un modelo matemático inspirado en el comportamiento biológico de las neuronas y en la estructura del cerebro.
Si bien las computadoras son muy rápidas en el procesamiento de la información, existen tareas muy complejas, como el reconocimiento y clasificación de patrones, que demandan demasiado tiempo y esfuerzo aún en las computadoras más potentes de la actualidad, pero que el cerebro humano es más apto para resolverlas, muchas veces sin aparente esfuerzo (considere por ejemplo el reconocimiento de un rostro familiar entre una multitud de otros rostros). El cerebro puede considerarse un sistema altamente complejo. Su unidad básica, la neurona, está masivamente distribuida con conexiones entre ellas (se calcula que hay aproximadamente 10 billones de neuronas en la corteza cerebral y 60 trillones de conexiones neuronales).
<p>Si bien hay distintos tipos de neuronas biológicas, en la siguiente figura se muestra un tipo particular que es muy común. </p>
<div style="margin: 0 auto; width:720px; color: gray;text-align:center;margin-bottom:20px;">
    <img src="img/neurona_biologica.png" width="720px" style="margin: 0 auto;"/><br/>
    Partes de una neurona biológica.
</div>

Vemos que la misma está compuesta por:</p>
<ul>
    <li>El cuerpo central, llamado soma, que contiene el núcleo celular</li>
    <li>Una prolongación del soma, el axón</li>
    <li>Una ramificación terminal, las dendritas</li>
    <li>Una zona de conexión entre una neurona y otra, conocida como sinapsis</li>
</ul>

<p>La función principal de las neuronas es la transmisión de los impulsos nerviosos. Estos viajan por toda la neurona comenzando por las dendritas hasta llegar a las terminaciones del axón, donde pasan a otra neurona por medio de la conexión sináptica. La manera en que respondemos ante los estímulos del mundo exterior y nuestro aprendizaje del mismo está directamente relacionado con las conexiones neuronales del cerebro, y las RNAs son un intento de emular este hecho.</p>

<h2>Modelo neuronal de McCulloch-Pitts</h2>
El primer modelo matemático de una <b>neurona artificial</b>, creado con el fin de llevar a cabo tareas simples, fué presentado en el año 1943 en un trabajo conjunto entre el psiquiatra y neuroanatomista <b>Warren McCulloch</b> y el matemático <b>Walter Pitts</b>. Un ejemplo de modelo neuronal con dos entradas x e y es representado en la siguiente figura:
<div style="margin: 0 auto; width:420px; color: gray;text-align:center;margin-bottom:20px;">
    <img src="img/neurona-mcculloch-p1.png" width="420px" style="margin: 0 auto;"/><br/>
    Modelo de McCulloch-Pitts para una neurona artificial.
</div>
El mismo consta de:
<ul>
    <li>Las entradas $x_1$ y $x_2$</li>
    <li>Los pesos sinápticos $w_1$ y $w_2$ correspondientes a cada entrada</li>
    <li>Un término aditivo $b$</li>
    <li>Una función de activación $f$</li>
    <li>Una salida $y$</li>
</ul>
Las entradas $x_1$ y $x_2$ son el estímulo que la neurona artificial recibe del entorno que la rodea, y la salida z es la respuesta a tal estímulo. La neurona se adapta al medio circundante y aprende de él modificando el valor de sus pesos sinápticos $w_1$ y $w_2$ y su término aditivo b. Estos son conocidos como los parámetros libres del modelo, pues los mismos pueden ser modificados y adaptados para realizar
una tarea determinada. En este modelo, la salida neuronal $y$ está dada por:
$$y = f(w_1 x_1 + w_2 x_2 + b)$$
La función de activación $f$ es seleccionada de acuerdo a la tarea realizada por la neurona.

<h2>La neurona (unidad)</h2>
<p>La neurona es la unidad funcional de los modelos de redes. Dentro de cada neurona, ocurren simplemente dos operaciones: la suma ponderada de sus entradas y la aplicación de una función de activación.</p>
<p>En la primera parte, se multiplica cada valor de entrada $x_i$ por su peso asociado $w_i$ y se suman junto con el <b>bías</b>. Este es el valor neto de entrada a la neurona. A continuación, este valor se pasa por una función, conocida como <b>función de activación</b>, que transforma el valor neto de entrada en un valor de salida.</p>
<p>Si bien el valor que llega a la neurona, multiplicación de los pesos por las entradas, siempre es una combinación lineal, gracias a la función de activación, se pueden generar salidas muy diversas. Es en la función de activación donde reside el potencial de los modelos de redes para aprender <b>relaciones no lineales</b>.</p>
<div style="margin: 0 auto; width:680px; color: gray;text-align:center;margin-bottom:20px">
    <img src="img/neurona.png" width="680px" style="margin: 0 auto;"/>
    Representación de una neurona.
</div>
La anterior ha sido una explicación intuitiva del funcionamiento de una neurona. Véase ahora una definición más matemática.
<p>El valor neto de entrada a una neurona es la suma de los valores que le llegan, ponderados por el peso de las conexiones, más el bias.</p>
$$entrada=\sum_{i=1}^n x_i w_i+b$$
En lugar de utilizando el sumatorio, esta operación suele representarse como el producto matricial, donde $X$ representa el vector de los valores de entrada y $W$ el vector de pesos.
$$entrada=XW+b$$
A este valor se le aplica una función de activación ($g$) que lo transforma en lo que se conoce como valor de activación ($a$), que es lo que finalmente sale de la neurona.
$$a=g(entrada)=g(XW+b)$$
Para la capa de entrada, donde únicamente se quiere incorporar el valor de los predictores, la función de activación es la unidad, es decir, sale lo mismo que entra. En la capa de salida, la función de activación utilizada suele ser la identidad para problemas de regresión y <i>soft max</i> para clasificación.

<h3>Ejemplo: Neurona MCP con 2 entradas</h3>
Para mostrar el funcionamiento básico de una neurona artificial (MCP) simularemos una compuerta lógica <b>OR</b> con dos entradas.
<div style="margin: 0 auto; width:700px; color: gray;text-align:center;margin-bottom:20px;margin-top:20px;">
    <img src="img/neurona_or.png" width="700px" style="margin:0 auto"/>
    Neurona con 2 entradas y tabla de verdad OR.
</div>
Sabemos que f=1, si
$$\sum_{i=1}^n x_i w_i>t$$
asi que sustituimos los valores $x_i$ de la tabla de verdad y obtenemos las siguientes desigualdades:
<table>
    <tr><td style="padding-right:120px;">$0(w_1) + 0(w_2) < t$</td><td>$0<t$</td></tr>
    <tr><td style="padding-right:120px;">$0(w_1) + 1(w_2) > t$</td><td>$w_2>t$</td></tr>
    <tr><td style="padding-right:120px;">$1(w_1) + 0(w_2) > t$</td><td>$w_1>t$</td></tr>
    <tr><td style="padding-right:120px;">$1(w_1) + 1(w_2) > t$</td><td>$w_1+w_2>t$</td></tr>
</table>

Se recomienda asignar valores a los pesos $w_1, w_2$ en el rango $[−1, +1]$. Posibles valores que cumplen con las desigualdades son:
$$t=0.5$$
$$w_1=0.7$$
$$w_2=0.7$$
Es decir, para el problema de la compuerta lógica OR de dos entradas $x_1$ y $x_2$, la neurona que simula el comportamiento de f está dada por los pesos $w_1$ y $w_2$ y el umbral $t$.
Si remplazamos los valores anteriores en el modelo MCP obtenemos:
$$0.7x_1+0.7x_2=0.5$$
$$x_2=\frac{0.5}{0.7}-x_1$$
<div>Esto representa la ecuación de una recta. Si $x_1=0$, obtenemos el punto de corte de la recta en $x_2=0.71$.</div>

    <img src="img\recta_mcp.png" width="360px" style="margin:0 auto;"/>
    Representación gráfica del MCP (clasificador lineal)
</div>

<h3>Aprendizaje: Regla Delta</h3>
Los pesos de una neurona se encuentran a través de las desigualdades. Es un proceso manual que si extendemos a 3, 4, 5, . . . , n entradas, el número de desigualdades crece exponencialmente.<br/>
Por lo anterior no resulta práctico el MCP sin un algoritmo que permita encontrar los pesos de manera automática. En 1962, Bernard Widrow y Ted Hoff propusieron la regla delta, un algoritmo para obtener los pesos de una neurona iterativamente, también conocida como regla de aprendizaje.
<div style="width:400px;margin:0 auto;color:#555555">
    <ol>
        <li>Selecciona una entrada de la tabla de verdad.</li>
        <li>Si se detecta un error, estimar que tan lejos está el MCP de la salida deseada.</li>
        <li>Ajustar los pesos ‘activos’ (p.e. cuando una entrada $x_i=1$) para remover la porción $d = \frac{(E + e)}{2}$ del error.</li>
        <li>Ir al paso 1, hasta que ninguna columna genere un error.</li>
    </ol>
    <b>Algoritmo 1: Regla delta</b>
</div>
El factor de corrección $d$ se define por el error $E$ y una constante de aprendizaje $e$ (similar al valor de temperatura en recocido simulado – simulated annealing).
<h4>Ejemplo: Regla delta</h4>
Obtendremos los pesos del problema anterior (OR) usando la regla delta. Los pesos iniciales para el ejemplo de la regla delta son: $w_1=-0.5, w_2 =0.2, t=0.8$. Los valores iniciales se definen de manera aleatoria (random), preferentemente en un rango de
$[−1, +1]$. El valor de la constante de aprendizaje se establece a $e=0.1$.
<h5>Iteración 1</h5>
<ol>
    <li>Selecciona una entrada de la tabla de verdad.<br/>
    $$x_1=0, x_2=0, f=0$$
    <span style="color:blue">$$0=0(-0.5)+0(0.2)<0.8$$</span></li>
</ol>
Sin error, continuamos con la siguiente entrada de la tabla de verdad.
<h5>Iteración 2</h5>
<ol>
    <li>Selecciona una entrada de la tabla de verdad.<br/>
    $$x_1=0, x_2=1, f=1$$
    <span style="color:red">$$0.2=0(-0.5)+1(0.2)>0.8$$</span></li>
    <li>Si se detecta un error, estimar que tan lejos está el MCP de la salida deseada.<br/>
        $E=0.6$ porque no se satisface la igualdad.</li>
    <li>Ajustar los pesos ‘activos’ (p.e. cuando una entrada $x_i=1$) para remover la porción del error.<br/>
    Los pesos ‘activos’, son $w_2$ y $t$, dado que $x_1=0$ y $x_2=1$. Se aplica el factor de corrección $d=\frac{(E + e)}{2}= \frac{(0.6+ 0.1)}{2} = 0.35$ a $w_2$ y $t$. En este caso, se le suma d a $w_2$ y se le resta d a t, por lo tanto los nuevos valores son:
        $$w_2=w_2+d =0.2+0.35=0.55$$ y
        $$t=t−d=0.8−0.35=0.45$$</li>
</ol>

<h5>Iteración 3</h5>
<ol>
    <li>Selecciona una entrada de la tabla de verdad.<br/>
    $$x_1=1, x_2=0, f=1$$
    <span style="color:red">$$-0.5=1(-0.5)+0(0.55)>0.45$$</span></li>
    <li>Si se detecta un error, estimar que tan lejos está el MCP de la salida deseada.<br/>
        $E=0.95$ porque no se satisface la igualdad.</li>
    <li>Ajustar los pesos ‘activos’ para remover la porción del error.<br/>
    Los pesos ‘activos’, son $w_1$ y $t$, dado que $x_1=1$ y $x_2=0$. Se aplica el factor de corrección $d=\frac{(E + e)}{2}= \frac{(0.95+ 0.1)}{2} =0.525$ a $w_1$ y $t$. En este caso, se le suma d a $w_1$ y se le resta d a t, por lo tanto los nuevos valores son:
        $$w_1=w_1+d =-0.5+0.525=0.025$$ y
        $$t=t−d=0.45−0.525=-0.075$$</li>
</ol>
<h5>Iteración 4</h5>
No hay error porque se cumple que $w_1+w_2>t$
<span style="color:blue">$$0.575=0.025+0.55>-0.075$$</span>


<h5>Iteración 5</h5>
<ol>
    <li>Selecciona una entrada de la tabla de verdad.<br/>
    $$x_1=0, x_2=0, f=0$$
    <span style="color:red">$$0=0(0.025)+0(0.55)<-0.075$$</span></li>
    <li>Si se detecta un error, estimar que tan lejos está el MCP de la salida deseada.<br/>
        $E=0.075$ porque no se satisface la igualdad.</li>
    <li>Ajustar los pesos ‘activos’ para remover la porción del error.<br/>
    El único peso ‘activo’, es $t$, dado que $x_1=0$ y $x_2=0$. Se aplica el factor de corrección $d=\frac{(E + e)}{2}= \frac{(0.075+ 0.1)}{2} =0.0875$ a $t$. En este caso, se le suma d a t, por lo tanto los nuevos valores son:
        $$t=t+d=-0.075+0.0875=0.0125$$</li>
</ol>
Todas las desigualdades se cumplen, es decir $E=0$ por lo que ya terminamos.
$$w_1=0.025$$
$$w_2=0.55$$
$$t=0.0125$$

<h2>Redes neuronales</h2>
<h3>Estructura de la red</h3>
<p>Las redes neuronales son modelos creados al ordenar operaciones matemáticas siguiendo una determinada estructura. La forma más común de representar la estructura de una red neuronal es mediante el uso de <b>capas</b> (layers), formadas a su vez por <b>neuronas</b> (unidades). Cada neurona, realiza una operación sencilla y está conectada a las neuronas de la capa anterior y de la capa siguiente mediante <b>pesos</b>, cuya función es regular la información que se <b>propaga</b> de una neurona a otra.</p>
<div style="margin: 0 auto; width:720px; color: gray;text-align:center;margin-bottom:20px;">
    <img src="img/estructura_basica.png" width="720px" style="margin: 0 auto;"/><br/>
    Representación de una red neuronal feed-forward (single-layer perceptron).
</div>

<p>La primera capa de la red neuronal (color verde) se conoce como capa de entrada o <b>input layer</b> y recibe los datos en bruto, es decir, el valor de los predictores. La capa intermedia (color azul), conocida como capa oculta o <b>hidden layer</b>, recibe los valores la capa de entrada, ponderados por los pesos (flechas grises). La última capa, llamada <b>output layer</b>, combina los valores que salen de la capa intermedia para generar la predicción.</p>
<p>Para facilitar la comprensión de la estructura de las redes, es útil representar una red equivalente a un modelo de regresión lineal.</p>
$$y=w_1x_1+...+w_dx_d+b$$
<div style="margin: 0 auto; width:240px; color: gray;font-size:0.7em;;text-align:center;margin-bottom:20px">
    <img src="img/vector.png" width="180px" style="margin: 0 auto;"/>
    Representación de una red neuronal equivalente a un modelo lineal con 4 predictores.
</div>

<p>Cada neurona de la capa de entrada representa el valor de uno de los predictores. Las flechas representan los coeficientes de regresión, que en términos de redes se llaman pesos, y la neurona de salida representa el valor predicho. Para que esta representación equivalga a la ecuación de un modelo lineal, faltan dos cosas:</p>
<ul>
    <li>El bias del modelo.</li>
    <li>Las operaciones de multiplicación y suma que combinan el valor de los predictores con los pesos del modelo.</li>
</ul>
La capa intermedia de una red tiene un valor de bias, pero suele omitirse en las representaciones gráficas. En cuanto a las operaciones matemáticas, es el elemento clave que ocurre dentro de las neuronas y conviene verlo con detalle.

<h2>Función de activación</h2>
Las funciones de activación controlan en gran medida que información se propaga desde una capa a la siguiente (<b>forward propagation</b>). Estas funciones convierten el valor neto de entrada a la neuronal, combinación de los input, pesos y bías, en un nuevo valor. Es gracias combinar funciones de activación no lineales con múltiples capas, que los modelos de redes son capaces de aprender relaciones no lineales.
<p>La gran mayoría de funciones de activación convierten el valor de entrada neto de la neurona en un valor dentro del rango (0, 1) o (-1, 1). Cuando el valor de activación de una neurona (salida de su función de activación) es cero, se dice que la neurona está inactiva, ya que no pasa ningún tipo de información a las siguientes neuronas. A continuación, se describen las funciones de activación más empleadas.</p>

<h3>Rectified linear unit (ReLU)</h3>
La función de activación ReLu aplica una transformación no lineal muy simple, activa la neurona solo si el input está por encima de cero. Mientras el valor de entrada está por debajo de cero, el valor de salida es cero, pero cuando es superior de cero, el valor de salida aumenta de forma lineal con el de entrada.
$$ReLU(x)=max(x,0)$$
De esta forma, la función de activación retiene únicamente los valores positivos y descarta los negativos dándoles una activación de cero.
<div style="margin: 0 auto; width:480px; color: gray;text-align:center;margin-bottom:20px">
    <img src="img/relu.png" width="480px" style="margin: 0 auto;"/><br/>
    Función de activación ReLU.
</div>
ReLU es con diferencia la función de activación más empleada, por sus buenos resultados en aplicaciones diversas. La razón de esto reside en el comportamiento de su derivada (gradiente), que es cero o constante, evitando así un problema conocido como <b>vanishing gradients</b> que limita la capacidad de aprendizaje de los modelos de redes.

<h3>Sigmoide</h3>
La función sigmoide transforma valores en el rango de ($-inf, +inf$) a valores en el rango (0, 1).
$$sigmoid(x)=\frac{1}{1+exp(-x)}$$
<div style="margin: 0 auto; width:480px; color: gray;text-align:center;margin-bottom:20px">
    <img src="img/sigmoide.png" width="480px" style="margin: 0 auto;"/><br/>
    Función de activación sigmoide.
</div>
Aunque la función de activación sigmoide se utilizó mucho en los inicios de los modelos de redes, en la actualidad, suele preferirse la función ReLU.
<p>Un caso en el que la función de activación sigmoide sigue siendo la función utilizada por defecto es en las neuronas de la capa de salida de los modelos de clasificación binaria, ya que su salida puede interpretarse como probabilidades.</p>

<h3>Tangente hiperbólica (Tanh)</h3>
La función de activación Tanh, se comporta de forma similar a la función sigmoide, pero su salida está acotada en el rango (-1, 1).
$$tanh(x)=\frac{1-exp(-2x)}{1+exp(-2x)}$$
<div style="margin: 0 auto; width:480px; color: gray;text-align:center;margin-bottom:20px">
    <img src="img/tanh.png" width="480px" style="margin: 0 auto;"/><br/>
    Función de activación Tangente hiperbólica.
</div>
Sin las funciones de activación, las redes neuronales solo pueden aprender relaciones lineales.
